```python
import pandas as pd
import numpy as np

dates = pd.date_range(
    start="2025-01-01",
    periods=12,
    freq="ME"
)

ndvi = np.random.uniform(
    0.2,
    0.9,
    12
)

df = pd.DataFrame({
    "Date": dates,
    "NDVI": ndvi
})

df
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    df["Date"],
    df["NDVI"],
    marker="o"
)

plt.title("NDVI Trend")
plt.xlabel("Date")
plt.ylabel("NDVI")
plt.grid(True)

plt.show()
```

```python
def classify_stress(ndvi):

    if ndvi >= 0.6:
        return "Healthy"

    elif ndvi >= 0.4:
        return "Moderate Stress"

    else:
        return "High Stress"


df["Stress_Level"] = df["NDVI"].apply(classify_stress)

df
```

```python
df["Stress_Level"].value_counts()
```

```python
stress_counts = df["Stress_Level"].value_counts()

plt.figure(figsize=(7,5))

stress_counts.plot(
    kind="bar"
)

plt.title("Climate Stress Distribution")
plt.xlabel("Stress Level")
plt.ylabel("Count")

plt.show()
```

```python
regions = [
    "Region_A",
    "Region_B",
    "Region_C",
    "Region_D",
    "Region_E"
]

region_ndvi = np.random.uniform(
    0.2,
    0.9,
    len(regions)
)

region_df = pd.DataFrame({
    "Region": regions,
    "NDVI": region_ndvi
})

region_df["Stress_Level"] = region_df["NDVI"].apply(classify_stress)

region_df
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    region_df["Region"],
    region_df["NDVI"]
)

plt.axhline(
    y=0.6,
    linestyle="--",
    label="Healthy Threshold"
)

plt.axhline(
    y=0.4,
    linestyle="--",
    label="Stress Threshold"
)

plt.title("Regional NDVI Comparison")
plt.ylabel("NDVI")
plt.legend()

plt.show()
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)

regions = [
    "Odisha",
    "Jharkhand",
    "West Bengal",
    "Chhattisgarh",
    "Telangana"
]

ndvi = np.random.uniform(0.2, 0.85, len(regions))

df = pd.DataFrame({
    "Region": regions,
    "NDVI": ndvi
})

df
```

```python
def classify_stress(ndvi):
    if ndvi >= 0.6:
        return "Healthy"
    elif ndvi >= 0.4:
        return "Moderate Stress"
    else:
        return "Severe Stress"

df["Stress_Level"] = df["NDVI"].apply(classify_stress)

df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(
    df["Region"],
    df["NDVI"]
)

plt.axhline(
    y=0.6,
    linestyle="--",
    label="Healthy Threshold"
)

plt.axhline(
    y=0.4,
    linestyle="--",
    label="Stress Threshold"
)

plt.title("Climate Stress Assessment Using NDVI")
plt.ylabel("NDVI")
plt.xlabel("Region")

plt.legend()

plt.show()
```

```python
df["Stress_Score"] = (1 - df["NDVI"]) * 100

df["Stress_Score"] = df["Stress_Score"].round(2)

df
```

```python
risk_df = df.sort_values(
    by="Stress_Score",
    ascending=False
)

risk_df
```

```python
import matplotlib.pyplot as plt

colors = []

for score in risk_df["Stress_Score"]:
    if score >= 60:
        colors.append("red")
    elif score >= 40:
        colors.append("orange")
    else:
        colors.append("green")

plt.figure(figsize=(10,5))

plt.bar(
    risk_df["Region"],
    risk_df["Stress_Score"],
    color=colors
)

plt.title("Regional Climate Stress Risk Ranking")
plt.ylabel("Stress Score")
plt.xlabel("Region")

plt.show()
```

```python
import folium

locations = {
    "Odisha": [20.9517, 85.0985],
    "Jharkhand": [23.6102, 85.2799],
    "West Bengal": [22.9868, 87.8550],
    "Chhattisgarh": [21.2787, 81.8661],
    "Telangana": [18.1124, 79.0193]
}

m = folium.Map(
    location=[22.5, 85.0],
    zoom_start=5
)

for _, row in df.iterrows():

    region = row["Region"]
    stress = row["Stress_Level"]
    score = row["Stress_Score"]

    if stress == "Healthy":
        color = "green"
    elif stress == "Moderate Stress":
        color = "orange"
    else:
        color = "red"

    folium.Marker(
        location=locations[region],
        popup=f"{region}<br>Stress Score: {score}",
        icon=folium.Icon(color=color)
    ).add_to(m)

m
```

```python
m.save("../data/processed/climate_stress_map.html")
print("Map saved successfully!")
```

```python
total_regions = len(df)

healthy = (df["Stress_Level"] == "Healthy").sum()

moderate = (df["Stress_Level"] == "Moderate Stress").sum()

severe = (df["Stress_Level"] == "Severe Stress").sum()

print("Climate Stress Assessment Summary")
print("-" * 40)
print("Total Regions:", total_regions)
print("Healthy Regions:", healthy)
print("Moderate Stress Regions:", moderate)
print("Severe Stress Regions:", severe)
```
